```python
import awkward as ak
import matplotlib.pyplot as plt
import numpy as np
import uproot

# Official public CERN ROOT tutorial dataset containing real CMS collision records
file_url = "http://root.cern/files/NanoAOD_DoubleMuon_CMS2011OpenData.root"

print("Opening CERN Open Data file...")
file = uproot.open(file_url)

# Access the 'Events' tree inside the file
tree = file["Events"]

# Extract muon kinematics and charge branches
muon_pt = tree["Muon_pt"].array()
muon_eta = tree["Muon_eta"].array()
muon_phi = tree["Muon_phi"].array()
muon_mass = tree["Muon_mass"].array()
muon_charge = tree["Muon_charge"].array()

# Filter for events that contain at least 2 muons
ev_filter = ak.num(muon_pt) >= 2
pt = muon_pt[ev_filter]
eta = muon_eta[ev_filter]
phi = muon_phi[ev_filter]
mass = muon_mass[ev_filter]
charge = muon_charge[ev_filter]

# Take the two leading (highest momentum) muons from each event
p1_pt, p2_pt = pt[:, 0], pt[:, 1]
p1_eta, p2_eta = eta[:, 0], eta[:, 1]
p1_phi, p2_phi = phi[:, 0], phi[:, 1]
p1_mass, p2_mass = mass[:, 0], mass[:, 1]
p1_q, p2_q = charge[:, 0], charge[:, 1]

# Require opposite-sign charges (e.g., mu+ and mu-)
os_filter = (p1_q * p2_q) == -1

p1_pt = p1_pt[os_filter]
p2_pt = p2_pt[os_filter]
p1_eta = p1_eta[os_filter]
p2_eta = p2_eta[os_filter]
p1_phi = p1_phi[os_filter]
p2_phi = p2_phi[os_filter]
p1_mass = p1_mass[os_filter]
p2_mass = p2_mass[os_filter]

# Convert to Cartesian 4-momentum components (px, py, pz, E)
px1 = p1_pt * np.cos(p1_phi)
py1 = p1_pt * np.sin(p1_phi)
pz1 = p1_pt * np.sinh(p1_eta)
E1 = np.sqrt(px1**2 + py1**2 + pz1**2 + p1_mass**2)

px2 = p2_pt * np.cos(p2_phi)
py2 = p2_pt * np.sin(p2_phi)
pz2 = p2_pt * np.sinh(p2_eta)
E2 = np.sqrt(px2**2 + py2**2 + pz2**2 + p2_mass**2)

# Calculate Invariant Mass of the dimuon system
E_tot = E1 + E2
px_tot = px1 + px2
py_tot = py1 + py2
pz_tot = pz1 + pz2
inv_mass = np.sqrt(
    np.maximum(0, E_tot**2 - (px_tot**2 + py_tot**2 + pz_tot**2))
)

# Plot the invariant mass spectrum
plt.figure(figsize=(10, 6))
plt.hist(
    inv_mass,
    bins=300,
    range=(0.2, 120),
    color="crimson",
    histtype="stepfilled",
    edgecolor="black",
    alpha=0.8,
)
plt.yscale("log")  # Log scale to easily see both low-mass and high-mass resonances
plt.xlabel("Invariant Mass $m_{\mu\mu}$ [GeV/$c^2$]", fontsize=12)
plt.ylabel("Events / Bin (Log Scale)", fontsize=12)
plt.title(
    "CMS Open Data: Dimuon Invariant Mass Spectrum ($J/\\psi$ and $Z$ Boson Peaks)",
    fontsize=14,
)
plt.grid(True, which="both", linestyle="--", alpha=0.5)

# Annotate peaks
plt.annotate(
    "J/\\psi (~3.1 GeV)",
    xy=(3.1, 100),
    xytext=(10, 500),
    arrowprops=dict(facecolor="black", shrink=0.05, width=1, headwidth=5),
)
plt.annotate(
    "Z Boson (~91 GeV)",
    xy=(91, 50),
    xytext=(70, 5000),
    arrowprops=dict(facecolor="black", shrink=0.05, width=1, headwidth=5),
)

plt.show()
```
